```python
import pandas as pd
from geopulse.store_validation import validate_store_data

test_stores = pd.DataFrame({
    "StoreID": ["S001", "S001", "", "S004", "S005"],
    "StoreName": ["Store A", "Store B", "", "Store D", "Store E"],
    "Lat": [28.65, 95.00, 28.63, "invalid", 28.61],
    "Long": [77.15, 77.18, 200.00, 77.12, "wrong"]
})

print("Invalid Store Dataset:")
display(test_stores)

print("\nStore Validation Results:")
validation_results = validate_store_data(test_stores)

for check, result in validation_results.items():
    print(f"{check}: {result}")
```
